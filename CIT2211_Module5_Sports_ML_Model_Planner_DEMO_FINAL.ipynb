## Step 2 — Quick Concept Review

Use this table to help students decide which learning type fits their sports problem.

| Learning Type | Best Used When | Sports Example | Output |
|---|---|---|---|
| **Supervised Learning** | You have labeled examples and want to predict an outcome | Predict injury risk using past injury labels | Low / Medium / High risk |
| **Supervised Learning** | You want to predict a number | Predict next-game points or race time | Numeric prediction |
| **Unsupervised Learning** | You want to find hidden groups or patterns | Group athletes by workload and recovery patterns | Clusters/groups |
| **Unsupervised Learning** | You want to detect unusual behavior | Identify abnormal performance patterns | Outlier/anomaly |

For most beginner assignments, if students say “predict,” they are usually describing **supervised learning**.


## Step 3 — Build the AI Planning Prompt

This prompt asks the AI assistant to help design the conceptual ML model. Students should review the output critically and revise it.


````python
def build_sports_ml_prompt(sport, goal, data_inputs):
    data_text = ", ".join(data_inputs)
    return f"""
You are helping a beginner sports analytics student design a conceptual machine learning model.

Sport: {sport}
Goal: {goal}
Data available: {data_text}

Please provide a clear, beginner-friendly answer with these sections:

1. Best Learning Type
   - Choose supervised or unsupervised learning.
   - Explain why it fits the goal.

2. Model Goal and Output
   - Explain what the model predicts or discovers.
   - Give an example output.

3. Useful Data Inputs / Features
   - Identify which data inputs are most useful.
   - Explain why they matter.

4. Conceptual Model Flow
   - Write the flow as: Input → Processing → Output → Human Decision.

5. Sports Impact
   - Explain how the model could improve performance, training, injury prevention, or strategy.

6. Limitations and Ethics
   - Identify data limitations, bias risks, privacy concerns, and why human review is required.

Keep the explanation practical and easy for a college student to understand.
""".strip()

prompt = build_sports_ml_prompt(SPORT, GOAL, DATA_INPUTS)
print(prompt)
````

## Step 6 — Ask Ollama for Planning Support

This cell tries to contact the local Ollama API. If it fails, the notebook will use the completed instructor sample output.


````python
import requests, json, textwrap

OLLAMA_URL = "http://localhost:11434/api/generate"

FALLBACK_SAMPLE_OUTPUT = """Best learning type: Supervised learning is the best fit because the goal is to predict a known outcome: fatigue risk before the next game. The model would learn from past examples where athletes had inputs such as minutes played, sprint count, heart rate, sleep hours, recovery score, turnovers, and days since last game, along with a label such as Low, Medium, or High fatigue risk.

Suggested model output: The model could return a fatigue risk category such as Low, Medium, or High. It could also provide a confidence level and a short explanation such as: high minutes played + high sprint count + low sleep + low recovery score.

Useful features: minutes played, sprint count, heart rate, sleep hours, recovery score, turnovers, recent workload, days since last game, travel schedule, and previous fatigue history.

Input → Processing → Output → Human Decision: Sports data is collected from wearable sensors, performance stats, and recovery logs. The model compares the current athlete profile against previous labeled examples. It outputs a fatigue-risk category. The coach or trainer reviews the result and decides whether to reduce workload, add recovery, adjust minutes, or request further evaluation.

Performance impact: This model could help coaches avoid overtraining, protect athletes from fatigue-related performance decline, and personalize recovery plans.

Limitations and ethics: The model may be wrong if data is incomplete, inaccurate, or missing context such as illness, stress, travel, or minor injury. Athlete privacy must be protected, and predictions should not be used to punish athletes. Human review is required before making training or playing-time decisions."""

def ask_ollama(prompt, model=MODEL_NAME, url=OLLAMA_URL, timeout=180):
    payload = {
        "model": model,
        "prompt": prompt,
        "stream": False
    }
    response = requests.post(url, json=payload, timeout=timeout)
    response.raise_for_status()
    return response.json().get("response", "")

try:
    ai_output = ask_ollama(prompt)
    print("✅ Ollama response received.
")
except Exception as e:
    ai_output = FALLBACK_SAMPLE_OUTPUT
    print("⚠️ Ollama was not available, so this notebook is using the instructor fallback sample output.
")
    print("Error summary:", str(e)[:300], "
")

print(ai_output)
````

## Step 8 — Generate a Simple Model Diagram

This diagram can be used in the assignment report or presentation.


````python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

steps = [
    "Sports Problem
Basketball fatigue risk",
    "Data Inputs
minutes, sprints, HR, sleep, recovery",
    "ML Processing
learns patterns from labeled examples",
    "Model Output
Low / Medium / High fatigue risk",
    "Human Decision
adjust workload or recovery"
]

fig, ax = plt.subplots(figsize=(14, 3.2))
ax.set_xlim(0, 14)
ax.set_ylim(0, 3)
ax.axis("off")

x_positions = [0.3, 3.0, 5.8, 8.7, 11.4]
box_w = 2.2
box_h = 1.3

for i, (x, text) in enumerate(zip(x_positions, steps)):
    box = FancyBboxPatch((x, 0.9), box_w, box_h, boxstyle="round,pad=0.08")
    ax.add_patch(box)
    ax.text(x + box_w/2, 1.55, text, ha="center", va="center", fontsize=9)
    if i < len(x_positions) - 1:
        arrow = FancyArrowPatch((x + box_w, 1.55), (x_positions[i+1], 1.55), arrowstyle="->", mutation_scale=15)
        ax.add_patch(arrow)

plt.title("Conceptual Sports ML Model Flow", fontsize=14)
plt.show()
````
